```python
customer_schema1= ['cust_id', 'cust_name', 'cust_city', 'cust_age']
orders_schema1 =['order_id', 'cust_id', 'products', 'price', 'order_date']
```

```python
customer_df1 = spark.createDataFrame(customer_data1, customer_schema1)
orders_df1 = spark.createDataFrame(orders_data1, orders_schema1)
```

```python
from pyspark.sql.functions import explode,count, col, lit, sum, avg, countDistinct, row_number, dense_rank, rank, lag, lead, when
from pyspark.sql.window import Window
```

```python
orders_df1.filter((col("cust_id") == 1) | (col("cust_id") == 2)).show(10,False)
```

```python
orders_cnt = orders_df1.withColumn("orders_cnt",count("order_id").over(Window.partitionBy("cust_id")))
```

```python
total_orders_per_cust_id = orders_cnt.select("cust_id","orders_cnt").distinct()
```

```python
customer_df1.show(5)
total_orders_per_cust_id.show(5)
```

```python
join_df = customer_df1.join(total_orders_per_cust_id, on='cust_id', how='left')
```

```python
join_df.filter(col("orders_cnt") == 1).show(3)
join_df.filter(col("orders_cnt") == 2).show(3)
join_df.filter(col("orders_cnt") == 3).show(3)
join_df.filter(col("orders_cnt").isNull()
               ).show(3)
```

```python
updated_join = join_df.withColumn("orders_cnt",when(col("orders_cnt").isNull(),lit(0)) \
                                 .otherwise(col("orders_cnt")))
```
