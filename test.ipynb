```python

from pyspark.sql.functions import explode, split, lower, regexp_replace

df = spark.read.text("/FileStore/tables/New_Text_Document.txt")

# Clean text: lowercase & remove punctuation
words_df = df.select(
    explode(
        split(
            lower(regexp_replace("value", "[^a-zA-Z ]", "")),
            " "
        )
    ).alias("word")
)

# Remove empty words
words_df = words_df.filter(words_df['word'] != "")

# Word count
word_count = words_df.groupBy("word").count()

# Display result
word_count.orderBy("count", ascending=False).show(truncate=False)
```
